## Section 3 Classify Logical Constraint Types

### Policy 1: Foundation Course Requirement
*   **IF–THEN:** If a new credit risk representative is being onboarded, then CRM101 must be assigned.
*   **Constraint Type:** Activation
*   **Explanation:** This is an Activation constraint because the implied decision to onboard a new representative (A) directly requires the assignment of CRM101 (B). CRM101 must always be selected.

### Policy 2: Advanced Analytics Prerequisite
*   **IF–THEN:** If CRM107 is assigned, then CRM106 must be assigned.
*   **Constraint Type:** Activation
*   **Explanation:** This is an Activation constraint because selecting CRM107 (A) necessitates the selection of CRM106 (B). CRM106 is a prerequisite for CRM107.

### Policy 3: Financial Statement & Regulatory Pairing
*   **IF–THEN:** If CRM102 is assigned, then CRM103 must be assigned.
*   **Constraint Type:** Activation
*   **Explanation:** This is an Activation constraint because choosing CRM102 (A) requires CRM103 (B) to be assigned alongside it, due to their interconnected nature.

### Policy 4: Specialized Analytics Exclusion
*   **IF–THEN:** If CRM106 is assigned, then CRM108 cannot be assigned.
*   **Constraint Type:** Suppression
*   **Explanation:** This is a Suppression constraint because the decision to assign CRM106 (A) actively prevents the assignment of CRM108 (B). They are mutually exclusive in the context of this policy.

### Policy 5: Client-Facing Skill Enhancement
*   **IF–THEN:** If CRM104 is assigned, then CRM105 must be assigned.
*   **Constraint Type:** Activation
*   **Explanation:** This is an Activation constraint because the assignment of CRM104 (A) triggers the requirement to also assign CRM105 (B) for effective client communication.

### Policy 6: Minimum Data Literacy
*   **IF–THEN:** At least one of CRM106 or CRM107 must be assigned.
*   **Constraint Type:** Either-or
*   **Explanation:** This is an Either-or constraint because it dictates that a representative must be assigned either CRM106, or CRM107, or both, to satisfy the minimum data literacy requirement.
*   **Ambiguity Clarification:** The phrase "at least one" clearly indicates that selecting both CRM106 and CRM107 is permissible. If the intention were "exactly one" (i.e., mutually exclusive choice), I would clarify with the policy owner to ensure the rule is not interpreted as allowing both.
